```python
def get_8bit_string(value):
    return format(value & 0xFF, '08b')

def logical_left_shift(binary_str, shift_amt):
    val = int(binary_str, 2)
    print(f"Original 8-bit number: {binary_str}")
    
    current_val = val
    for i in range(1, shift_amt + 1):
        current_val = current_val << 1
        print(f"Step {i} (Shift Left 1):  {get_8bit_string(current_val)}")
        
    final_result = get_8bit_string(current_val)
    print(f"Final 8-bit result:   {final_result}\n")
    return final_result

def logical_right_shift(binary_str, shift_amt):
    val = int(binary_str, 2)
    print(f"Original 8-bit number: {binary_str}")
    
    current_val = val
    for i in range(1, shift_amt + 1):
        current_val = current_val >> 1
        print(f"Step {i} (Shift Right 1): {get_8bit_string(current_val)}")
        
    final_result = get_8bit_string(current_val)
    print(f"Final 8-bit result:   {final_result}\n")
    return final_result

print("--- Operation A: Logical Left Shift ---")
logical_left_shift("10110110", 3)

print("--- Operation B: Logical Right Shift ---")
logical_right_shift("11010100", 3)
```

```python
def to_twos_complement_8bit(val):
    return format(val & 0xFF, '08b')

def from_twos_complement_8bit(binary_str):
    val = int(binary_str, 2)
    if binary_str[0] == '1':
        val -= 256
    return val

def add_binary_8bit(bin1, bin2):
    result = []
    carry = 0
    for i in range(7, -1, -1):
        bit1 = int(bin1[i])
        bit2 = int(bin2[i])
        
        total_sum = bit1 + bit2 + carry
        bit_out = total_sum % 2
        carry = total_sum // 2
        
        result.append(str(bit_out))
    
    result.reverse()
    return "".join(result), carry

def solve_expression(num1, num2, label):
    print("=" * 60)
    print(f" Arithmetic Expression {label}: {num1} + ({num2})")
    print("=" * 60)
    
    bin_num1 = to_twos_complement_8bit(num1)
    bin_num2 = to_twos_complement_8bit(num2)
    
    print(f"Operand 1: {num1:4d}  ->  Binary (2's complement): {bin_num1}")
    print(f"Operand 2: {num2:4d}  ->  Binary (2's complement): {bin_num2}")
    print("-" * 60)
    
    bin_result, final_carry = add_binary_8bit(bin_num1, bin_num2)
    dec_result = from_twos_complement_8bit(bin_result)
    
    print(f"Binary Operation:")
    print(f"   {bin_num1}")
    print(f" + {bin_num2}")
    print(f"   {'-'*8}")
    print(f"   {bin_result}  (8-bit truncated result)")
    print("-" * 60)
    
    sign1 = bin_num1[0]
    sign2 = bin_num2[0]
    sign_res = bin_result[0]
    
    overflow = (sign1 == sign2) and (sign1 != sign_res)
    
    print(f"Final Binary Result : {bin_result}")
    print(f"Final Decimal Result: {dec_result}")
    print(f"Expected True Mathematical Sum: {num1 + num2}")
    
    if overflow:
        print("\n[ALERT] Arithmetic Overflow Detected!")
        print("Reason: The true result exceeds the 8-bit signed range [-128 to +127].")
        if num1 > 0 and num2 > 0:
            print("Type: Positive Overflow (result wrapped around to a negative value)")
        else:
            print("Type: Negative Overflow (result wrapped around to a positive value)")
    else:
        print("\n[INFO] No Arithmetic Overflow. Result is valid.")
    print("=" * 60 + "\n")

if __name__ == "__main__":
    solve_expression(120, 62, "a")
    solve_expression(-120, -62, "b")
```
